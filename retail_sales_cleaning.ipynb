```python
import pandas as pd
df =pd.read_csv("retail_store_sales.csv")
df.head(25)
```

```python
df["Price Per Unit"]=df["Price Per Unit"].fillna(df["Total Spent"]/df["Quantity"])
```

```python
df[df["Category"]=="Electric household essentials"][["Item","Price Per Unit"]].sort_values("Item")
```

```python
df.head(40)
```

```python
category_to_item ={
    "Milk Products": "milk",
    "Patisserie":"PAT",
    "Butchers": "BUT",
    "Beverages": "BEV",
    "Furniture": "FUR",
    "Food": "FOOD",
    "Electric household essentials": "EHE",
    "Computers and electric accessories":"CEA"
}
```

```python
df["Item"]=df["Item"].fillna(df["Category"].map(category_to_item))
```

```python
df["Price Per Unit"].isna().sum()
```

```python
df[["Quantity","Total Spent"]]=df[["Quantity","Total Spent"]].fillna(df[["Quantity","Total Spent"]].median())
```

```python
mask = (
    df["Quantity"].isna() &
    df["Price Per Unit"].notna()&
    df["Total Spent"].notna()
)
```

```python
(df["Quantity"].isna()&df["Total Spent"].isna()).sum()
```

```python
df.loc[mask,["Quantity","Price Per Unit","Total Spent"]].head(20)
```

```python
df["Price Per Unit"]=df["Price Per Unit"].fillna(df.groupby("Category")["Price Per Unit"].transform("median"))
```

```python
df.isnull().sum()
```

```python
df.loc[df["Item"].isna(),["Category",'Item']]["Category"].value_counts(dropna=False)
```

```python
df["Item"]=df["Item"].replace("milk","MILK")
```

```python
df["Discount Applied"]=df["Discount Applied"].fillna("Unknown")
```
